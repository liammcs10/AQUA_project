```python
import pickle
import matplotlib.pyplot as plt

with open("results_test.pickle", "rb") as file:
    data = pickle.load(file)

print(data[:][0])


N = len(data)
CAC = np.zeros(N)
CP = np.zeros(N)

for n, row in enumerate(data):
    CAC[n] = row[0]
    CP[n] = row[1]

plt.plot(CAC, c = 'blue', label = 'CAC')
plt.plot(CP, c = 'red', label = 'CP')
plt.legend()

```
