```python
"""
Test for the presence of chaos in the IB model with autapse.

Simulate the IB neuron with many different autapse strengths (f) to a fixed driving
current. Around the bifurcation point, we hope to see some changes in indicators of
chaos...

"""

# local imports
import aqua
from aqua.batchAQUA_general import batchAQUA
from aqua.AQUA_general import AQUA
from aqua.stimulus import *
from aqua.plotting_functions import *

# packages
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
from scipy.fft import fft, ifft, fftfreq


IB = {'name': 'IB', 'C': 150, 'k': 1.2, 'v_r': -75, 'v_t': -45, 'v_peak': 50,
     'a': 0.01, 'b': 5, 'c': -56, 'd': 130, 'e': 0.2, 'f': 0., 'tau': 2.}
```

```python
T = 50000
dt = 0.1
N_iter = int(T/dt)

f_values = np.arange(200, 300, 10)
N_neurons = len(f_values)
print(f_values)

I_HEIGHT = 680
I_inj = I_HEIGHT * np.ones((N_neurons, N_iter))

params = []

for f in f_values:
    temp = IB.copy()
    temp['f'] = f
    params.append(temp)

df = pd.DataFrame(params)

x_start = np.full((N_neurons, 3), fill_value = np.array([-60, 0, 0]))
t_start = np.zeros(N_neurons)

batch = batchAQUA(params)
batch.Initialise(x_start, t_start)

X, t, spikes = batch.update_batch(dt, N_iter, I_inj)
```
